```python
import pandas as pd
merged = pd.read_csv("../data/processed/P01116_merged.csv")
res = pd.read_csv("../data/processed/P01116_residues.csv")
sub = merged[merged["label"].isin(["pathogenic", "benign"])]
p = sub[sub.label == "pathogenic"]
b = sub[sub.label == "benign"]
print(len(p), "pathogenic,", len(b), "benign")
```

```python
print("All residues:")
print(res["ss"].value_counts(normalize=True).round(2))
print("\nPathogenic variants:")
print(p["ss"].value_counts(normalize=True).round(2))
```

```python
from scipy.stats import mannwhitneyu
for col in ["rsa", "plddt"]:
    print(col, "p =", round(mannwhitneyu(p[col], b[col]).pvalue, 4))
```

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import Image

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
for a, col in zip(ax, ["rsa", "plddt"]):
    a.boxplot([b[col], p[col]], tick_labels=["benign", "pathogenic"])
    a.set_title(col)
plt.tight_layout()
plt.savefig("../results/figures/pathogenic_vs_benign.png", dpi=200)
plt.close(fig)
Image("../results/figures/pathogenic_vs_benign.png")
```
